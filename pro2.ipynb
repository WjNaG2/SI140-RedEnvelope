### 2.2数据分析
观察三张图可以发现，无论红包的金额以及个数，每个人抢到的金额分步都呈现出相似性：每个人的抢到的红包金额的期望相近，且越晚抢的人所抢到的金额的方差越大。因此我们猜测，微信红包的机制和红包金额以及红包个数无关。在接下来的研究过程中不再考虑红包金额和红包个数的影响，选择60元红包15人抢数据作为研究对象。
